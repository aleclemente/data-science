### Preparação dos dados

Nessa fase iremos remover as linhas / instâncias que o atributo está faltando, remover a coluna ID caso esteja presente e separar o dados de treino e teste.

```python
import numpy as np


# -------------------------------------------------------------------------
# Passô 1: Remover os dados de X e y em conjunto onde tem algum atributo ausente
# -------------------------------------------------------------------------

# 1. Padronizar eventuais placeholders (como '?') para NaN em X
X_aux = X.replace(['?', '', ' '], np.nan)

# 2. Identificar linhas que possuem pelo menos um valor nulo em qualquer coluna de X
linhas_com_nulos = X_aux.isna().any(axis=1)

# 3. Remover as linhas correspondentes de X e y
X_limpo = X[~linhas_com_nulos].copy()
y_limpo = y.loc[~linhas_com_nulos].copy()

# 4. Verificar se ainda restou algum valor vazio/nulo em qualquer coluna
ainda_tem_nulos = X_limpo.isna().any().any() or X_limpo.isin(['?', '', ' ']).any().any()

print(f"Ainda existe algum valor vazio em X? {ainda_tem_nulos}")
print(f"Total de instâncias limpas: {X_limpo.shape[0]}")

# -------------------------------------------------------------------------
# Passo 2: Remover a coluna de id 'Sample_code_number'
# -------------------------------------------------------------------------
print(f"\n--- 2. Removendo a coluna de ID ---")

# Remove a coluna de ID de X, caso ela esteja presente
if 'Sample_code_number' in X_limpo:
    X_limpo = X_limpo.drop(columns=['Sample_code_number'])
else:
    print("A coluna de ID não está presente no conjunto de dados.")
```

```python
from sklearn.model_selection import train_test_split

# -------------------------------------------------------------------------
# Passo 3: Separar dados de treino e de testes
# -------------------------------------------------------------------------
print("\n--- 3. Separando em treino e teste (ex: 80% treino, 20% teste) ---")

X_train, X_test, y_train, y_test = train_test_split(
    X_limpo,
    y_limpo,
    test_size=0.2,
    random_state=42,
    stratify=y_limpo # Mantém a proporção das classes (benigno/maligno) nos dois conjuntos
)

print(f"Tamanho do conjunto de treino (X_train): {X_train.shape}")
print(f"Tamanho do conjunto de teste (X_test): {X_test.shape}")
print(f"Tamanho do conjunto de treino (y_train): {y_train.shape}")
print(f"Tamanho do conjunto de teste (y_test): {y_test.shape}")

print("\nPrimeiras 5 linhas de X_train:")
display(X_train.head())
print("\nPrimeiras 5 linhas de y_train:")
display(y_train.head())
```

### Modelagem

Calcular os centróides médios de cada atributo de treino e usar para classificar os dados de teste e depois trazer a acúrária.

```python
import numpy as np
from sklearn.metrics import accuracy_score, confusion_matrix

print("\n--- 4. Classificador Baseado em Centróides Médios ---")

# As classes alvo são 2 (benigno) e 4 (maligno)
class_labels = [2, 4]

# 1. Calcular os centróides para cada classe nos dados de treino
centroids = {}
for label in class_labels:
    # Filtrar X_train para obter apenas as amostras da classe atual
    # Certifique-se que y_train tem a mesma indexação que X_train para o filtro
    # Usamos .iloc para garantir o alinhamento de índices após o reset_index
    X_train_class = X_train[y_train['Class'] == label]
    centroids[label] = X_train_class.mean(axis=0)
    print(f"Centróide da Classe {label}:\n{centroids[label].head()}\n")

# 2. Função para fazer previsões com base nos centróides
def predict_centroid_classifier(X_data, centroids):
    predictions = []
    for _, row in X_data.iterrows():
        distances = {}
        for label, centroid in centroids.items():
            # Calcular a distância Euclidiana entre a amostra e o centróide
            distances[label] = np.linalg.norm(row - centroid)

        # A classe prevista é aquela cujo centróide é o mais próximo
        predicted_label = min(distances, key=distances.get)
        predictions.append(predicted_label)
    return np.array(predictions).reshape(-1, 1)

# 3. Fazer previsões nos dados de teste
y_pred = predict_centroid_classifier(X_test, centroids)

```

### Avaliação

```python
# 4. Avaliar o desempenho do classificador

# Acurácia geral
overall_accuracy = accuracy_score(y_test, y_pred)
print(f"Acurácia Geral do Classificador Centróide: {overall_accuracy:.4f}\n")

# Acurácia por classe
# Obter a matriz de confusão
cm = confusion_matrix(y_test, y_pred, labels=class_labels)

# Calcular a acurácia por classe a partir da matriz de confusão
class_accuracies = {}
for i, label in enumerate(class_labels):
    tp = cm[i, i] # True Positives para a classe atual
    fn = cm[i, (i + 1) % len(class_labels)] # False Negatives (assumindo 2 classes)

    # Se for uma classe binária, a matriz de confusão tem 2x2. tp = cm[i,i], total_real_class = cm[i,:].sum()
    total_real_class = cm[i, :].sum()
    if total_real_class > 0:
        class_accuracies[label] = tp / total_real_class
    else:
        class_accuracies[label] = 0.0 # Evitar divisão por zero se a classe não tiver instâncias reais no teste

print("Acurácia por Classe:")
for label, accuracy in class_accuracies.items():
    print(f"  Classe {label}: {accuracy:.4f}")

print("\nMatriz de Confusão:\n", cm)
```
