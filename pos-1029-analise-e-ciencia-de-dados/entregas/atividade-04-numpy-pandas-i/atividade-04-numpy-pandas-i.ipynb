## Questão 8 — Coluna derivada com NumPy (1,0 ponto)

Crie a coluna `faixa` com `np.where`, segundo a regra:

- `nota_final` >= 85 → `"Excelente"`
- 70 <= `nota_final` < 85 → `"Bom"`
- 60 <= `nota_final` < 70 → `"Regular"`
- `nota_final` < 60 → `"Insuficiente"`

**Cuidado:** quem está sem nota deve ficar com `NaN` em `faixa`, e não cair
na última categoria. Mostre `df["faixa"].value_counts(dropna=False)`.

````python
print("## Coluna faixa com np.where \n")

df["faixa"] = np.where(
    df["nota_final"].isna(),
    "NaN", # Using None instead of np.nan to avoid DTypePromotionError when mixing with strings
    np.where(
        df["nota_final"] >= 85,
        "Excelente",
        np.where(
            df["nota_final"] >= 70,
            "Bom",
            np.where(df["nota_final"] >= 60, "Regular", "Insuficiente"),
        ),
    ),
)

# Mostra a contagem de valores para a nova coluna 'faixa', incluindo NaNs
print(df["faixa"].value_counts(dropna=False))
````

## Questão 10 — Uma pergunta respondida com dados (1,0 ponto)

Responda, com código **e** com uma frase de interpretação:

> **Os estudantes do turno noturno têm desempenho diferente dos demais?**

Calcule a nota média por turno, a quantidade de estudantes em cada turno e
a taxa de aprovação por turno. Depois escreva, em uma célula de texto, o que
os números mostram — e **uma limitação** dessa conclusão.

> Lembre-se: os dados são sintéticos. Avalia-se o raciocínio, não a conclusão.

````python
print("## Quantidade de estudantes por turno \n")
contagem_turno = df["turno"].value_counts(dropna=False)
print(contagem_turno)
print("\n----------\n")

print("## Nota média por turno \n")
media_nota_turno = df.groupby("turno")["nota_final"].mean().round(2)
print(media_nota_turno)
print("\n----------\n")

print("##  Taxa de aprovação por turno \n")
taxa_aprovacao_turno = (
    df.groupby("turno")["situacao"]
    .apply(lambda x: (x == "Aprovado").mean() * 100)
    .round(2)
)
print(taxa_aprovacao_turno)
print("\n----------\n")

print("## Exibindo os resultados consolidados \n")
resumo_turnos = pd.DataFrame(
    {
        "Quantidade de Estudantes": contagem_turno,
        "Nota Média": media_nota_turno,
        "Taxa de Aprovação (%)": taxa_aprovacao_turno,
    }
)
print(resumo_turnos)
print("\n----------\n")
````

_Sua resposta escrita:_

Os números mostram que os estudantes do turno noturno possuem uma nota média e uma taxa de aprovação ligeiramente inferiores às dos turnos matutino e vespertino, sugerindo que o período em que o estudante frequenta as aulas pode estar associado a um menor desempenho acadêmico geral.

Limitação da Conclusão:

A principal limitação dessa análise é que ela estabelece apenas uma correlação, e não uma relação de causalidade, ignorando outras variáveis.
